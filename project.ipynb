Model Building

1. Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
def linear_regression(x_train, y_train, x_test, y_test):
  model = LinearRegression()
  model.fit(x_train, y_train)
  y_predicted = model.predict(x_test)

  rmse = np.sqrt(mean_squared_error(y_test, y_predicted))
  r2 = r2_score(y_test, y_predicted)

  print(f"Root Mean Squared Error : {rmse}")
  print(f"R-squared score : {r2}")

  return model,y_predicted
```

```python
model, y_predicted = linear_regression(x_train, y_train, x_test, y_test)
```

2. Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
def decision_tree(x_train, y_train, x_test, y_test):
  model = DecisionTreeRegressor()
  model.fit(x_train, y_train)
  y_predicted = model.predict(x_test)

  rmse = np.sqrt(mean_squared_error(y_test, y_predicted))
  r2  = r2_score(y_test, y_predicted)

  print(f"Root Mean Squared Error : {rmse}")
  print(f"R-squared Score : {r2}")

  return model, y_predicted
```

```python
model, y_predicted = decision_tree(x_train, y_train, x_test, y_test)
```

3. Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
def random_forest(x_train, y_train, x_test, y_test, n_estimators = 2):
  model = RandomForestRegressor()
  model.fit(x_train, y_train)
  y_predicted = model.predict(x_test)

  rmse = np.sqrt(mean_squared_error(y_test, y_predicted))
  r2 = r2_score(y_test, y_predicted)

  print(f"Root Mean Squared Error : {rmse}")
  print(f"R-squared Score : {r2}")

  return model, y_predicted
```

```python
model, y_predicted = random_forest(x_train, y_train, x_test, y_test, n_estimators=2)

# didn't execute, since it took a long time to execute. 
```

Support Vector Machine

```python
from sklearn.svm import SVR
```

for SVM only one target is required.

```python
y_stator_winding = y_train['stator_winding']
y_stator_tooth = y_train['stator_tooth']
y_stator_yoke = y_train['stator_yoke']
y_pm = y_train['pm']

y_stator_winding_test = y_test['stator_winding']
y_stator_tooth_test = y_test['stator_tooth']
y_stator_yoke_test = y_test['stator_yoke']
y_pm_test = y_test['pm']
```

```python
def SVR_model(x_train, y_stator_winding, x_test, y_stator_winding_test, sample_size=5000):
    x_sample = x_train.sample(n=sample_size, random_state=42)
    y_sample = y_stator_winding.loc[x_sample.index]
    model = SVR()
    model.fit(x_sample, y_sample)
    y_predicted = model.predict(x_test)

    rmse = np.sqrt(mean_squared_error(y_stator_winding_test, y_predicted))
    r2 = r2_score(y_stator_winding_test, y_predicted)

    print(f"Root Mean Squared Error : {rmse}")
    print(f"R-squared Score : {r2}")

    return model, y_predicted  
```

```python
model, y_predicted = SVR_model(x_train, y_stator_winding, x_test, y_stator_winding_test)
```
